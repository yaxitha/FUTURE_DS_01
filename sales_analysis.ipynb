# Sales Data Analysis Project
This notebook performs basic sales data analysis and visualization.

```

import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
file_path = 'dataset/sales_data.csv'
df = pd.read_csv(file_path, encoding='latin1')

# Display first rows
df.head()
```

```

# Convert Order Date column
if 'Order Date' in df.columns:
    df['Order Date'] = pd.to_datetime(df['Order Date'])

# KPI Calculations
sales_column = 'Sales'
profit_column = 'Profit'

print("Total Sales:", df[sales_column].sum())
print("Total Profit:", df[profit_column].sum())
print("Total Orders:", df['Order ID'].nunique())
```

```

# Sales Trend
sales_trend = df.groupby(df['Order Date'].dt.month)['Sales'].sum()

plt.figure(figsize=(8,5))
sales_trend.plot()
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.savefig('sales_trend.png')
plt.show()
```

```

# Top Products
top_products = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))
top_products.plot(kind='bar')
plt.title('Top Products')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.tight_layout()
plt.savefig('top_products.png')
plt.show()
```

```

# Regional Sales
regional_sales = df.groupby('Region')['Sales'].sum()

plt.figure(figsize=(8,5))
regional_sales.plot(kind='bar')
plt.title('Regional Sales')
plt.xlabel('Region')
plt.ylabel('Sales')
plt.tight_layout()
plt.savefig('regional_sales.png')
plt.show()
```

```

# Category Profit
category_profit = df.groupby('Category')['Profit'].sum()

plt.figure(figsize=(8,5))
category_profit.plot(kind='bar')
plt.title('Category Profit')
plt.xlabel('Category')
plt.ylabel('Profit')
plt.tight_layout()
plt.savefig('category_profit.png')
plt.show()
```
